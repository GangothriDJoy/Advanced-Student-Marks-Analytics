```python
%%writefile substring.l
%{
#include <string.h>
%}
%option noyywrap
%%
"abc"   { printf("ABC"); }
.       { printf("%s", yytext); }
\n      { putchar('\n'); }
%%
int main(){
    printf("Enter the input string: ");
    yylex();
    printf("\n");
    return 0;
}
```

```python
%%bash
apt-get update
apt-get install -y flex
```

```python
%%writefile counting.l
%{
int line_count=0;
int word_count=0;
int char_count=0;
%}

%option noyywrap

%%

[^ \t\n]+ {word_count++;} {char_count+=yyleng;}
\n {line_count++;} {char_count++;}
. {char_count++;}

%%

int main(){
printf("Enter the imput text: ");
fflush(stdout);
yylex();
printf("DEBUG: yylex() returned.\n");
printf("Total Line count=%d\n", line_count);
printf("Total Word count=%d\n", word_count);
printf("Total Character count=%d\n", char_count);
}
```

```python
!flex counting.l
!gcc -o counting lex.yy.c
```

```python
%%bash
echo -e "You think darkness is your ally?\nYou merely adopted it.\nI was born in it.\nMolded by it." > input.txt
cat input.txt
```

```python
%%bash
./counting < input.txt
```
